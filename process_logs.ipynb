```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
selected_columns = df.iloc[:, 5:245]

selected_columns = df.iloc[:, 160:240].replace({
    "True": 1,
    "False": 0,
    True: 1,
    False: 0
})

heatmap_df = selected_columns.apply(pd.to_numeric, errors="coerce")
plt.figure(figsize=(64, 32))
sns.heatmap(
    heatmap_df,
    cmap="coolwarm",
)

plt.show()
```

```python
#I want to plot column and average value of each column bar plot
plt.figure(figsize=(10, 5))
plt.bar(selected_columns.columns, [selected_columns[col].mean() for col in selected_columns.columns])
plt.xlabel("Columns")
plt.ylabel("Average Value")
plt.title("Average Values of Each Column")
plt.show()
```
